```python
segment_counts = customers["Customer_Segment"].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=segment_counts.index,
    y=segment_counts.values
)

plt.title("Customer Distribution by Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")

plt.show()
```

### Visualization 1 - Customer Distribution by Segment

**What it shows:**  
The distribution of FinTrust's 1,500 customers across the four customer segments.

**Why it matters:**  
The size of each customer segment affects how overall transaction activity should be interpreted. A larger segment may naturally generate more transactions because it contains more customers.

**Business Insight:**  
Everyday customers are the largest customer segment, with 711 customers (47.4%), while SME customers are the smallest, with 216 customers (14.4%). Therefore, segment-level transaction patterns should be considered alongside the size of each customer segment.

```python
transaction_type_counts = transactions["Transaction_Type"].value_counts()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=transaction_type_counts.index,
    y=transaction_type_counts.values
)

plt.title("Transaction Activity by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=30)

plt.show()
```

### Visualization 2 - Transaction Activity by Transaction Type

**What it shows:**  
The number of FinTrust transactions across the six transaction types.

**Why it matters:**  
Transaction frequency helps identify which banking services customers use most often. This can help FinTrust understand where transaction activity is concentrated.

**Business Insight:**  
Transfers are the most frequently used transaction type, with 3,549 transactions, followed by Card Purchases with 3,033. Airtime/Data has the lowest transaction volume at 1,185. The high volume of Transfers indicates that this transaction type represents an important area of customer activity for FinTrust.

```python
transaction_value = (
    transactions.groupby("Transaction_Type")["Amount_NGN"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=transaction_value.index,
    y=transaction_value.values
)

plt.title("Total Transaction Value by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Total Transaction Value (NGN)")
plt.xticks(rotation=30)

plt.show()
```

### Visualization 3 - Total Transaction Value by Transaction Type

**What it shows:**  
The total monetary value of transactions across the six transaction types.

**Why it matters:**  
Transaction frequency does not necessarily represent transaction value. Comparing both measures helps FinTrust understand which transaction types contribute the greatest financial value.

**Business Insight:**  
Transfers generate the highest total transaction value at approximately ₦237.9 million. Deposits generate approximately ₦131.0 million despite having fewer transactions than Card Purchases. This shows that transaction frequency and transaction value provide different perspectives on customer activity and should be considered together.

```python
channel_counts = transactions["Channel"].value_counts()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=channel_counts.index,
    y=channel_counts.values
)

plt.title("Transaction Activity by Channel")
plt.xlabel("Channel")
plt.ylabel("Number of Transactions")

plt.show()
```

### Visualization 4 - Transaction Activity by Channel

**What it shows:**  
The number of transactions completed through each FinTrust banking channel.

**Why it matters:**  
Understanding channel usage helps FinTrust identify where customers are most active and where operational performance may have the greatest impact.

**Business Insight:**  
The Mobile App is the dominant transaction channel, with 5,102 transactions (42.52% of all transactions). POS follows with 2,393 transactions, while USSD has the lowest activity at 889 transactions. This indicates that digital channels, particularly the Mobile App, play an important role in FinTrust's transaction activity.

```python
status_counts = transactions["Transaction_Status"].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=status_counts.index,
    y=status_counts.values
)

plt.title("Transaction Status Distribution")
plt.xlabel("Transaction Status")
plt.ylabel("Number of Transactions")

plt.show()
```

### Visualization 5 - Transaction Status Distribution

**What it shows:**  
The distribution of FinTrust transactions across successful, failed, reversed and pending statuses.

**Why it matters:**  
Transaction status provides an indication of how consistently transactions are completed successfully and helps identify the scale of unsuccessful transaction outcomes.

**Business Insight:**  
The majority of transactions are successful, with 10,856 transactions (90.47%) completed successfully. Failed transactions account for 630 transactions (5.25%), while reversed and pending transactions represent smaller proportions. This indicates that overall transaction completion is high, while failed, reversed and pending transactions remain areas for operational monitoring.

```python
risk_by_type = (
    transactions.groupby("Transaction_Type")["Risk_Review_Flag"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=risk_by_type.index,
    y=risk_by_type.values
)

plt.title("Risk Review Rate by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Risk Review Rate (%)")
plt.xticks(rotation=30)

plt.show()
```

### Visualization 6 - Risk Review Rate by Transaction Type

**What it shows:**  
The percentage of transactions marked for risk review across each transaction type.

**Why it matters:**  
Comparing risk-review rates across transaction types helps FinTrust identify where review activity is more concentrated and where further investigation may be useful.

**Business Insight:**  
Transfer transactions have the highest risk-review rate at 28.49%, followed by Cash Withdrawals at 25.31%. Bill Payments have the lowest rate at 12.81%. The variation suggests that review activity differs across transaction types and may warrant transaction-specific investigation. However, the Risk_Review_Flag is a synthetic review indicator and should not be interpreted as confirmed fraud or actual banking risk.

```python
risk_by_international = (
    transactions.groupby("International_Transaction")["Risk_Review_Flag"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=risk_by_international.index,
    y=risk_by_international.values
)

plt.title("Risk Review Rate: International vs Domestic Transactions")
plt.xlabel("International Transaction")
plt.ylabel("Risk Review Rate (%)")

plt.show()
```

### Visualization 7 - Risk Review Rate: International vs Domestic Transactions

**What it shows:**  
The percentage of international and domestic transactions marked for risk review.

**Why it matters:**  
Comparing international and domestic transactions helps identify whether review activity is distributed differently across transaction categories.

**Business Insight:**  
International transactions have an observed risk-review rate of 36.88%, compared with 18.88% for domestic transactions. This represents an 18.00 percentage-point difference. The pattern may warrant further investigation into the characteristics of international transactions. However, the Risk_Review_Flag is a synthetic review indicator and should not be interpreted as confirmed fraud or actual banking risk.

```python
customer_activity = (
    transactions.groupby("Customer_ID")
    .size()
    .reset_index(name="Transaction_Count")
)

customer_activity = customer_activity.merge(
    customers[["Customer_ID", "Digital_Engagement_Score"]],
    on="Customer_ID",
    how="left"
)

customer_activity["Engagement_Quartile"] = pd.qcut(
    customer_activity["Digital_Engagement_Score"],
    q=4,
    labels=["Q1 - Lowest", "Q2", "Q3", "Q4 - Highest"]
)
```

```python
engagement_activity = (
    customer_activity
    .groupby("Engagement_Quartile", observed=True)["Transaction_Count"]
    .mean()
    .reset_index()
)

engagement_activity
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=engagement_activity,
    x="Engagement_Quartile",
    y="Transaction_Count"
)

plt.title("Average Transactions by Digital Engagement Level")
plt.xlabel("Digital Engagement Level")
plt.ylabel("Average Transactions per Customer")

plt.show()
```

### Visualization 8 - Average Transactions by Digital Engagement Level

**What it shows:**  
The average number of transactions per customer across four digital-engagement quartiles, from the lowest to the highest engagement level.

**Why it matters:**  
This analysis examines whether customer digital engagement is associated with transaction activity, addressing a key Week 1 analytical question.

**Business Insight:**  
Customers in the highest digital-engagement quartile record the highest average transaction activity at approximately 8.26 transactions per customer. However, transaction activity does not increase consistently across all four engagement groups. This suggests a modest observed association rather than a clear or causal relationship between digital engagement and transaction activity.
